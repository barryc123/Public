## 1.1 Simple Linear Regression

$Y \approx \beta_0 + \beta_1 X$
* X is a single predictor variable
* Assumes there is a relationship between X and Y
* 'We are regressing Y onto X'
* $\beta_0$ and $\beta_1$ are unknown constants, they are coefficients/parameters
<br><br><br>

$\hat{y} \approx \hat{\beta_0} + \hat{\beta_1}x$ 
<br>
* '$\hat{}$' values are estimate~s
* $\hat{y}$ is a prediction of $y$ on the basis $X=x$
* $\hat{\beta_0}$ and $\hat{\beta_1}$ are modle coefficient estimates

### 1.1.1 Least Squares Criterion - Estimating the coefficients
<br>

* $\hat{y_i} = \hat{\beta_0} + \hat{\beta_1}x_i$ is the prediction for $Y$ based on the the $i^{th}$ value of $X$
* $\epsilon_i = y_i - \hat{y_i}$ is the $i^{th}$ residual, the difference between the $i^{th}$ observed response and the $i^{th}$ predicted value
* We define the residual sum of squares (**RSS**) as:

$ RSS = \epsilon_1^2 + \epsilon_2^2 + \epsilon_3^2 + ... + \epsilon_n^2$<br>
or equivalently: <br>
$ RSS = (y_1 - \hat{\beta_0} - \hat{\beta_1}x_1)^2 + (y_2 - \hat{\beta_0} - \hat{\beta_1}x_2)^2 + ... + (y_n - \hat{\beta_0} - \hat{\beta_1}x_n)^2$
<br><br>
* Least squares chooses $\hat{\beta_0}$ and  $\hat{\beta_1}$ to minimise the RSS:

$
\hat{\beta}_1
=
\frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}
     {\sum_{i=1}^{n} (x_i - \bar{x})^2}
$

$
\hat{\beta}_0
=
\bar{y} - \hat{\beta}_1 \bar{x}
$


### 1.1.2 Assessing the accuracy of the coefficient estimates 

* We assume the true relationship between $X$ and $Y$ takes the form $Y = f(X) + \epsilon$, for some unknown function $f$, where $\epsilon$ is a mean-zero random error term
* If $f$ is approximated by a linear function we can write this relationship as:
<br>$ Y = \beta_0 + \beta_1X + \epsilon$ <br>
where $\beta_0$ is the intercept, $\beta_1$ is the slope
