# Linear Regression 

## 1.1 Simple Linear Regression

$Y \approx \beta_0 + \beta_1 X$
* X is a single predictor variable
* Assumes there is a relationship between X and Y
* 'We are regressing Y onto X'
* $\beta_0$ and $\beta_1$ are unknown constants, they are coefficients/parameters
<br><br><br>

$\hat{y} \approx \hat{\beta_0} + \hat{\beta_1}x$ 
<br>
* '$\hat{}$' values are estimate~s
* $\hat{y}$ is a prediction of $y$ on the basis $X=x$
* $\hat{\beta_0}$ and $\hat{\beta_1}$ are modle coefficient estimates

### 1.1.1 Least Squares Criterion - Estimating the coefficients
<br>

* $\hat{y_i} = \hat{\beta_0} + \hat{\beta_1}x_i$ is the prediction for $Y$ based on the the $i^{th}$ value of $X$
* $\epsilon_i = y_i - \hat{y_i}$ is the $i^{th}$ residual, the difference between the $i^{th}$ observed response and the $i^{th}$ predicted value
* We define the residual sum of squares (**RSS**) as:

$ RSS = \epsilon_1^2 + \epsilon_2^2 + \epsilon_3^2 + ... + \epsilon_n^2$<br>
or equivalently: <br>
$ RSS = (y_1 - \hat{\beta_0} - \hat{\beta_1}x_1)^2 + (y_2 - \hat{\beta_0} - \hat{\beta_1}x_2)^2 + ... + (y_n - \hat{\beta_0} - \hat{\beta_1}x_n)^2$
<br><br>
* Least squares chooses $\hat{\beta_0}$ and  $\hat{\beta_1}$ to minimise the RSS:

$
\hat{\beta}_1
=
\frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}
     {\sum_{i=1}^{n} (x_i - \bar{x})^2}
$

$
\hat{\beta}_0
=
\bar{y} - \hat{\beta}_1 \bar{x}
$


<br><br>
### 1.1.2 Assessing the accuracy of the coefficient estimates 

* We assume the true relationship between $X$ and $Y$ takes the form $Y = f(X) + \epsilon$, for some unknown function $f$, where $\epsilon$ is a mean-zero random error term
* If $f$ is approximated by a linear function we can write this relationship as:
<br>$ Y = \beta_0 + \beta_1X + \epsilon$ <br>
where $\beta_0$ is the intercept, $\beta_1$ is the slope
<br><br>

* $SE(\hat{u})$, the standard error of $\hat{u}$ (the sample mean) is a measure of how far off the single estimate of $\hat{u}$ is from the population $u$

$\operatorname{Var}(\hat{\mu}) = \operatorname{SE}(\hat{\mu})^2 = \frac{\sigma^2}{n}$
<br> where $\sigma$ is the standard deviation of each of the realisations $y_i$ of $Y$

$
\operatorname{SE}(\hat{\beta}_0)^2
=
\sigma^2
\left[
\frac{1}{n}
+
\frac{\bar{x}^2}{\sum_{i=1}^{n} (x_i - \bar{x})^2}
\right]
$

$
\operatorname{SE}(\hat{\beta}_1)^2
=
\frac{\sigma^2}{\sum_{i=1}^{n} (x_i - \bar{x})^2}
$
<br><br> ,where $\sigma^2 = Var(\epsilon)$
<br><br>
* The estimate of $\sigma$ is known as the residual standard error **(RSE)**
<br>$\operatorname{RSE} = \sqrt{\frac{\operatorname{RSS}}{n - 2}}$<br>
* When $\sigma^2$ is estimated from the data we should write it as $\hat{SE}(\hat{\beta_1})$ but we drop the hat for convenience
* Standard errors can be used to create **confidence intervals**
* For linear regssion the 95% confidence interval for $\beta_1$ takes the form $\hat{\beta_1} \pm 2SE(\hat{\beta_1})$
* There is a 95% chance the interval $\left[\, \hat{\beta}_1 - 2 \cdot \operatorname{SE}(\hat{\beta}_1),\;
\hat{\beta}_1 + 2 \cdot \operatorname{SE}(\hat{\beta}_1) \,\right]$  will contain the true value of $\beta_1$
<br><br>

* Standard errors can be used to perform hypothesis tests on the coefficients:
<br><br>$H_0$: There is no relationship between $X$ and $Y$
<br><br>versus the alternative hypothesis<br>
<br>$H_1$: There is some relationship between $X$ and $Y$

* Mathematically this corresponds to:
<br><br>$H_0 : \beta_1 = 0$
<br><br>
$H_1 : \beta_1 \neq 0$
* To test the null hypothesis we need to determine whether $\hat{\beta_1}$, our estimate for $\beta_1$ is sufficiently far from zero thta we can be confident that $\beta_1$ is non-zero
* How far is enough? That depends on the $SE(\hat{\beta_1})$

<br><br>
* In practice we compute a **t-statistic**:
<br>$t=\frac{\hat{\beta}_1 - 0}{\operatorname{SE}(\hat{\beta}_1)}$
<br> which measures the number of standard deviations $\hat{\beta_1}$ is away from 0
* We use this to calculate the **p-value**
* A small p-value indicates it is unlikely to observe such a substantial association between the predictor and the response due to chance, in the absence of any real association between the predictor and the response
* We reject the null hypothesis, that is we declare a relationship between $X$ and $Y$, if the p-value is small enough
* Typical p-value cutoffs are 5% and 1%

<br><br>
## 1.1.3 Assessing the accuracy of the model 

* Once we have rejected the null hypopthesis in favour of the alternative hypothesis, we want to quantify the extent to which the model fits the data
* The quality of a linear regression is usually assessed using:
1. **Residual Standard Error (RSE)**
2. $R^2$ **Statistic**


### Residual Standard Error (RSE)
* RSE is an estimation of the standard deviation of $\epsilon$
* It's the averge amount the response will deviate from the true regression line
<br><br>
$
\operatorname{RSE}
=
\sqrt{\frac{1}{n - 2} \, \operatorname{RSS}}
=
\sqrt{
\frac{1}{n - 2}
\sum_{i=1}^{n} (y_i - \hat{y}_i)^2
}
$
<br><br>
e.g. If RSE=326 and the avergae value of y is 1000, then the model will eb off by 326 on average, which is 326/1000 = 23%
<br><br>
* RSE is considered as a measure of the lack of fit of the model

<br><br>
### $R^2$ Statistic
* RSE is measured in the units of the model so it is not always clear what it means
* $R^2$ provides an alternative measure of fit
* $R^2$ takes the form of a proportion - the proportion of variance explained - and so it always takes a value between 0 and 1, independent of the scale of Y
<br><br>
$R^2 = \frac{\operatorname{TSS} - \operatorname{RSS}}{\operatorname{TSS}} = 1 - \frac{\operatorname{RSS}}{\operatorname{TSS}}$
<br><br>
where $\operatorname{TSS} = \sum_{i=1}^{n} (y_i - \bar{y})^2$
<br><br>
* TSS measures the total variance in the response Y
* RSS measures the amount of variability left unexplained after performing the regression
* So $R^2$ measures the proportion of variability in Y that can be explained using X
* Value closer to 1 means it has greater explanatory power, closer to 0 means less explanatory power

## 1.2 Multiple Linear Regression
<br>
$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \cdots + \beta_p X_p + \epsilon$
<br><br>
$\beta_j$ is the avergae effect on $Y$ of a one unit increase in $X_j$, holding all other predictors fixed 
<br><br>

### 1.2.1 Estimating the Regression Coefficients
<br><br>
$\hat{y} = \hat{\beta}_0 + \hat{\beta}_1 x_1 + \hat{\beta}_2 x_2 + \cdots + \hat{\beta}_p x_p$
<br><br>
* Again we choose $\beta_0, \beta_1, ..., \beta_p$ to minimise the sum of squared residuals
<br><br>
$
\begin{aligned}
\mathrm{RSS}
&= \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 \\
&= \sum_{i=1}^{n} \left(y_i - \hat{\beta}_0 - \hat{\beta}_1 x_{i1} - \hat{\beta}_2 x_{i2} - \cdots - \hat{\beta}_p x_{ip}\right)^2
\end{aligned}
$
<br><br>


### 1.2.2 Some Important Considerations
1. Is at least one of the predictors $X_1, X_2, ... , X_p$ useful in predicting the response?
2. Do all predictors help to explain $Y$, or is only a subset of the predictors useful?
3. How well does the model fit the data?

#### 1. Is there a relationship between the response and the predictors?
* In multiple regression we need to check if all the regression coeffiicents are zero, i.e. $\beta_1 = \beta_2 = ... = \beta_p = 0$
* **Null Hypothesis:**  $\beta_1 = \beta_2 = ... = \beta_p = 0$
<br>versus the alternative<br>
* **Alternative Hypothesis:** At least one $\beta_j$ is non-zero
* This hypothesis is performed by computing the **F-Statistic**
<br><br>
$F = \dfrac{(\mathrm{TSS} - \mathrm{RSS})/p}{\mathrm{RSS}/(n - p - 1)}$
<br><br>
where in the same as simple linear regression:
<br>
$\mathrm{TSS} = \sum (y_i - \bar{y})^2$
<br>
$\mathrm{RSS} = \sum (y_i - \hat{y}_i)^2$
<br><br>
* If the linear model assumptions are correct then: $E\{\mathrm{RSS}/(n - p - 1)\} = \sigma^2$
* And provided $H_0$ is true: $E\{(\mathrm{TSS} - \mathrm{RSS})/p\} = \sigma^2$
* Hence, when there is no relationship between the response and the predictors we expect an F value close to 1
* If $H_1$ is true, we expect F to be greater than 1
<br><br>

### 2. Deciding on important variables
* If the F-statistic shows at least one of the predictors is related to the response, the next step is to determine which one(s) are related to it
* This step is referred to as variable selection
* Ideally we would like to try all possible combinations of models using subsets of the predictors
* We would then determine which model is best by using one of the following statistics:
<br>1.  Mallows $C_p$
<br>2. Akaike Information Criteria (AIC)
<br>3. Bayesion Information Criterion (BIC)
<br>4. Adusted $R^2$
* Unfortunately this is too computationally expensive in most cases. There are $2^p$ models that contain subsets of the predictors. So if $p=2$ there are $2^2=4$ models, but if $p=30$ there are $2^{30} = 1,073,741,824$ models!

* So we must choose another method for varibale selection:
    * **Forward Selection**
        * Begin with the null model -> the model that contains an intercept and no predictors
        * Fit $p$ simple linear regressions and add to the null model the one that has the lowest $RSS$
        * Then add to that model the variable that results in the lowest RSS for the new two-variale model
        * Continue the approach until some stopping rule is satisfied
    <br><br>
    *  **Backward Selection**
        * Start with all variables in the model and remove the variable with the largets p-value (the least statistically significant variable)
        * The new (p-1) variable model is fit, and again remove the variable with the largest p-value
        * Continue until some stopping rule is satisfied, e.g. all remaining variables have a p-value below some threshold
      <br><br>
    *  **Mixed Selection**
        *  Combination of forward and backward selection
        *  Start with forward selection and continue to add variables one by one
        *  But p-values for variables may become larger as new variables are added to the model. So if the p-value for one of the variables rises above a certain level, remove it
        *  Continue the forward and backward steps until all variables in the model have sufficiently low p-value, and all variables outside the model would have a large p-value if added
* Note: backward selction cannot be used if p > n
<br><br> 
### 3. Model Fit
* Two common measures of model fit are $RSE$ and $R^2$, the fraction of variance explained
* But in multiple linear regression $R^2$ will always increase when more variables are added to the model, even if they are only weakly associated with the response
* So $RSE$ and $R^2$ need to be considered together

## 1.3 Non-Linear Relationships 
* Linear regression assumes a linear relationship between the predictors and the response
* But the true relationship may be non-linear
* We can use **polynomial regression** to extend the linear model to accommodate non-linear relationships
<br><br>
$\hat{y} = \hat{\beta_0} + \hat{\beta_1}x + \hat{\beta_2}x^2 + \epsilon$
<br><br>
* It is still a linear model!
